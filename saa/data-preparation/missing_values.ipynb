# Missing values

A missing value is a cell that should hold a value and doesn't. Most models can't take one, so every missing value has to be removed, filled, or left to a model that accepts it. Which of these is right depends less on how many values are missing than on **why** they are missing.

## Unknown, or not applicable

Two checks come before any decision, and both are fixed rules applied when the table is built ([Where data preparation ends](where_preparation_ends.ipynb) says why):

- **Placeholders.** A missing value may be stored as something else: 0, −1, 999, `"?"`, `"unknown"`, an empty string. pandas sees a value; it must become a real missing value (`NaN`, [a special floating-point value](../../pia/python-data-essentials/numpy/numpy1.ipynb#nan)) first. [Pima diabetes](../supervised-learning/basic-workflows/pima_diabetes.ipynb) has insulin levels of 0 that are measurements never taken.
- **Not applicable.** A value may be absent because it doesn't exist, not because it is unknown: the date of the last change of owner, for a car that never had one. That absence is information, a value of its own ("no change of owner"), and is encoded as such, not filled as if it were unknown.

What remains are values that exist but weren't recorded. They are the subject of this notebook.

## A small table

Six cars, with the mileage of two and the fuel of one unknown:

```
import numpy as np
import pandas as pd

cars = pd.DataFrame({
    "km": [98_500, np.nan, 187_000, 12_900, np.nan, 120_750],
    "fuel": ["diesel", "petrol", None, "electric", "petrol", "hybrid"],
    "price": [9_800, 11_500, 3_900, 24_900, 14_200, 4_500],
})
cars
```

pandas marks every gap with `NaN`, in the text column too. `NaN` is a floating-point value, and integers have no such value, so the mileages, whole numbers with two gaps, are stored as floats: 98500.0. `NaN` isn't even equal to itself (`np.nan == np.nan` is `False`), so gaps are found with `isna`, never with `==`:

```
cars.isna().sum()
```

Dropping every row with a missing value:

```
cars.dropna()
```

Three cells out of eighteen were missing, and half of the rows are gone, with the values they had in their complete columns. Dropping every column with a missing value instead:

```
cars.dropna(axis="columns")
```

Only the price is left. Filling each gap with an estimate instead, which is called **imputation**: the mileage with the median of the known mileages, the fuel with the most frequent fuel:

```
cars.fillna({"km": cars["km"].median(), "fuel": cars["fuel"].mode()[0]})
```

Both cars of unknown mileage get 109,625 km, the middle of the four known values. Yet their prices, 11,500 and 14,200, are among the highest, and in this table the dearer cars have the lower mileages: the other columns say something about the missing value that the median ignores. Filling from the other columns is possible too, in a deep dive at the end of this notebook.

## The Titanic passengers

The passengers of the Titanic, from OpenML: one row per passenger, and whether they survived as the target. The features kept are the class (`pclass`), the sex, the age, the numbers of siblings or spouses (`sibsp`) and of parents or children (`parch`) aboard, the fare, the port where they embarked and the cabin. Three kinds of column are left out: those recorded after the sinking, the lifeboat and the body number, which would leak the answer; the identifiers, name and ticket; and the free text, home and destination ([Kinds of features](complementary/kinds_of_features.ipynb) explains each kind). The gaps are explored on the training rows only, like everything that may shape the model ([Exploring a dataset](complementary/exploring_a_dataset.ipynb)):

```
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split

passengers = fetch_openml("titanic", version=1, as_frame=True, data_home="data").frame
features = ["pclass", "sex", "age", "sibsp", "parch", "fare", "embarked", "cabin"]
X = passengers[features]
y = passengers["survived"].astype(int)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
X_train.isna().sum()
```

Three columns have gaps in the training rows: the cabin, for most passengers; the age, for about one in five; the fare, for one. Are the passengers without an age like the others? The share of passengers without an age, in each class:

```
X_train["age"].isna().groupby(X_train["pclass"]).mean().round(2)
```

And the survival rate of the passengers with and without a recorded age:

```
y_train.groupby(X_train["age"].isna().rename("age missing")).mean().round(2)
```

The age is missing far more often in third class, and the passengers without an age survived less often. Had the gaps been a random sample of the passengers, which is called **missing completely at random** (MCAR), the shares by class and the survival rates would differ only by chance. They differ by far more: whether the age is missing says something about the passenger, and about the target.

The cabin, the same way:

```
X_train["cabin"].isna().groupby(X_train["pclass"]).mean().round(2)
```

```
y_train.groupby(X_train["cabin"].isna().rename("cabin missing")).mean().round(2)
```

The cabin is known for most first-class passengers and almost no one else, and the passengers with a known cabin survived about twice as often. Whether the cabin is missing says more than the cabin itself.

The port of embarkation has no gap in the training rows, but the whole passenger list has two. Checking which rows they are is a check of the data, which may look at every row:

```
passengers.loc[passengers["embarked"].isna(), ["name", "pclass", "ticket", "fare", "cabin"]]
```

Two first-class passengers on the same ticket and in the same cabin: one missing record, not a pattern. (The Encyclopedia Titanica, the source of the dataset, says that both boarded at Southampton.) Both fell among the test rows. New data can have gaps in a column that had none in the training data, so whatever handles the gaps must be ready for a gap in any column, not only those with gaps today.

## The options, and what each assumes

Three questions choose among the options: which model will learn from the data, whether the gaps carry information, and whether new rows will have gaps too.

| Option | What it assumes | When it fits |
| --- | --- | --- |
| Leave it to the model | The model can learn where rows with a gap should go | Models that accept missing values: scikit-learn's trees, random forests and histogram gradient boosting, and XGBoost |
| Fill with a statistic | A typical value is a harmless stand-in for the missing one | Models that need a number in every cell |
| Flag the gap | Being missing may itself be informative | Combined with a fill, whenever the gaps are not MCAR |
| Drop the column | The column isn't worth what it costs | A column missing almost everywhere; keep a flag if its absence says something |
| Drop the rows | The rows with gaps are like the others (MCAR), and new rows will have no gaps | Very few gaps that new data will never have |

Filling each gap from similar rows or from the other columns, and comparing the options by how well a model predicts with each, are deep dives at the end of this notebook.

## Leaving it to the model

Some models accept missing values. scikit-learn's decision trees, random forests and histogram gradient boosting (`HistGradientBoostingClassifier` and `HistGradientBoostingRegressor`), and XGBoost ([XGBoost](../supervised-learning/algorithms/tree-based/xgboost.ipynb)), learn at every split of every tree on which side the rows with a missing value should go, choosing the side that reduces the error most. The gaps are not filled at all, and their being missing can be used directly.

Every other model needs a number in every cell: linear models, k-nearest neighbours, support vector machines and neural networks can't take a missing value. For them, the gaps are filled, or the columns or rows that have them are dropped.

## Filling with a statistic

scikit-learn's `SimpleImputer` learns a value for each column with `fit` (the mean, the median, the most frequent value, or a constant) and fills the gaps with it with `transform`. Like every step that learns from the data, it is fitted on the training rows only ([Where data preparation ends](where_preparation_ends.ipynb) says why). The median of the training ages:

```
from sklearn.impute import SimpleImputer

median_imputer = SimpleImputer(strategy="median").set_output(transform="pandas")
median_imputer.fit(X_train[["age"]])
median_imputer.statistics_
```

The median rather than the mean, because it is not pulled by extreme values ([Outliers and capped values](complementary/outliers.ipynb)). Every passenger without an age gets 28: over 200 passengers now share exactly the same age, and the filled ages vary less than real ones, since every filled value sits in the middle.

For categories, `strategy="most_frequent"` fills with the most common value: Southampton, for the port of embarkation.

## Flagging the gap

Since the passengers without an age survived less often, the fact that the age is missing is information worth keeping. `add_indicator=True` adds a column that says, for each row, whether the value was filled:

```
flagging_imputer = SimpleImputer(strategy="median", add_indicator=True).set_output(transform="pandas")
flagging_imputer.fit_transform(X_train[["age"]]).head()
```

The model can now treat a filled 28 differently from a real 28. Flagging is cheap, and worth trying whenever the gaps are not MCAR.

A model that uses a gap as information depends on how the data are collected. If the age became a required field, the flag would always be 0, and what the model learned from it would no longer apply. The same holds for a tree model that uses the gaps directly.

## Dropping a column, keeping what its absence says

The cabin is missing for most passengers, so as a value it is nearly useless. But its absence is informative, so dropping the column would lose that. A column that says whether the cabin was recorded, the same flag built by hand, keeps it:

```
X_train_cabin = X_train.assign(cabin_known=X_train["cabin"].notna()).drop(columns="cabin")
X_train_cabin.head()
```

This is a fixed rule (it learns nothing from the data), so it could be applied to every row before the split. The deck, the first letter of the cabin, would be a feature for the passengers whose cabin is known ([Creating features](complementary/creating_features.ipynb)).

## Dropping rows

Dropping the passengers without an age, and comparing what is left with the whole training set:

```
with_age = X_train["age"].notna()
pd.DataFrame({
    "all training passengers": [y_train.mean(), (X_train["pclass"] == 3).mean()],
    "only those with an age": [y_train[with_age].mean(), (X_train.loc[with_age, "pclass"] == 3).mean()],
}, index=["survival rate", "share in third class"]).round(3)
```

The passengers who are left survived more often and are less often in third class: they are no longer a sample of all the passengers. If whether the age is missing depends on survival itself, and not only on features the model sees, such as the class, the predictions of a model trained on them can be biased: a **selection bias**.

There is a second, more practical problem: the model will also be asked about new rows with missing values. Rows can be dropped from the training data, not from the data it will be used on. Dropping rows fits only gaps that new data will never have, such as a value that is always known in practice and is missing here by accident.

## Why a value is missing *(deep dive)*

Statisticians distinguish three mechanisms:

- **Missing completely at random (MCAR).** Whether a value is missing has nothing to do with any value, recorded or not: a page of a form lost, a sensor that fails for a few minutes whatever it measures. The rows with the value missing are a random sample of all the rows.
- **Missing at random (MAR).** Whether a value is missing depends only on other columns that *are* recorded. In a survey where women report their weight less often than men, the weight is MAR if, among women and among men separately, who reports it is a matter of chance. The name is confusing: the gaps are not random overall, only once the recorded columns are taken into account.
- **Missing not at random (MNAR).** Whether a value is missing depends on something not recorded, often the missing value itself: people with very high debts are less likely to declare them; a thermometer that can't read above 100 °C records nothing when it is hotter.

The data can show that gaps are not MCAR: if the rows with a gap differ from the others in their recorded columns, the gaps are not a random sample. The data can't tell MAR from MNAR, because the evidence for MNAR is in the values that are missing. That last step needs knowledge of how the data were collected.

The ages of the Titanic passengers are not MCAR: their gaps depend at least on the class, which is recorded. If the recorded columns explain them completely, the age is MAR; if they also depend on the passengers' own ages, it is MNAR. The data can't tell which: that needs knowing how the passenger list was compiled.

The three mechanisms come from statistical estimation, where the aim is a quantity, such as an average or a regression coefficient, and a fill that ignores the mechanism can bias it. No fill recovers MNAR values: the information needed is not in the data. A predictive model has a different aim, and the gap itself can be one of its inputs: even MNAR gaps can be used, through a flag or by a model that accepts missing values.

## Filling from similar rows, or from the other columns *(deep dive)*

Two imputers use the recorded columns to fill each gap differently:

- `KNNImputer` finds, for each passenger with a gap, the *k* passengers most similar in the other columns, and takes the average of their values. Similarity is a distance, so the columns must be scaled first ([Scaling numerical features](feature_scaling.ipynb)).
- `IterativeImputer` predicts each column with gaps from the others with a regression model, and repeats the process for several rounds, each one using the values filled in the previous round. scikit-learn still marks it as experimental: it must be enabled with a special import.

Both work on numbers only. The ages each one gives to a few passengers, next to the median:

```
from sklearn.experimental import enable_iterative_imputer  # noqa: F401 (enables the import below)
from sklearn.impute import IterativeImputer, KNNImputer
from sklearn.preprocessing import StandardScaler

numeric = ["pclass", "age", "sibsp", "parch", "fare"]
scaler = StandardScaler().set_output(transform="pandas").fit(X_train[numeric])
knn_imputer = KNNImputer(n_neighbors=5).set_output(transform="pandas").fit(scaler.transform(X_train[numeric]))
iterative_imputer = IterativeImputer(random_state=0).set_output(transform="pandas").fit(X_train[numeric])

missing_age = X_train[X_train["age"].isna()].head(6)
knn_ages = knn_imputer.transform(scaler.transform(missing_age[numeric]))["age"] * scaler.scale_[1] + scaler.mean_[1]
missing_age[["pclass", "sibsp", "parch", "fare"]].assign(
    median=median_imputer.transform(missing_age[["age"]])["age"],
    k_nearest_neighbours=knn_ages,
    iterative=iterative_imputer.transform(missing_age[numeric])["age"],
).round(1)
```

The median gives everyone 28. The two other imputers give each passenger an age that depends on their class, their fare and the relatives travelling with them, which is what MAR gaps call for, and they disagree with each other, by more than ten years for some passengers. The `KNNImputer` works on the scaled columns, so its ages are scaled back to years to compare them.

No imputer recovers the real ages; each gives an estimate under its own assumptions, and all of them make the filled values less varied than real ones.

## Comparing the options *(deep dive)*

Each option can be judged by how well a model predicts with it. Every fill that learns from the data goes inside the model's pipeline, so that cross-validation fits it again on the training part of each fold and the validation part never shapes it ([Data leakage](../supervised-learning/ml-fundamentals/data_leakage.ipynb)).

Five ways of handling the gaps: the median, the median with a flag, the nearest neighbours and the iterative fill, each inside a pipeline with a logistic regression, and gradient boosting, given the gaps as they are. The categories are filled with their most frequent value and one-hot encoded ([Encoding categories](categorical_encoding.ipynb)); the cabin is reduced to whether it is known. Each pipeline is measured by 10-fold cross-validation on the training rows:

```
from sklearn.compose import make_column_transformer
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder

X_train_model = X_train.assign(cabin_known=X_train["cabin"].notna()).drop(columns="cabin")
categories = make_pipeline(SimpleImputer(strategy="most_frequent"), OneHotEncoder(handle_unknown="ignore"))


def logistic_with(numeric_steps):
    preprocessing = make_column_transformer(
        (make_pipeline(*numeric_steps), numeric),
        (categories, ["sex", "embarked"]),
        remainder="passthrough",
    )
    return make_pipeline(preprocessing, LogisticRegression(max_iter=1000))


candidates = {
    "median": logistic_with([SimpleImputer(strategy="median"), StandardScaler()]),
    "median and flag": logistic_with([SimpleImputer(strategy="median", add_indicator=True), StandardScaler()]),
    "nearest neighbours": logistic_with([StandardScaler(), KNNImputer(n_neighbors=5)]),
    "iterative": logistic_with([IterativeImputer(random_state=0), StandardScaler()]),
    "gradient boosting, gaps left": make_pipeline(
        make_column_transformer((OneHotEncoder(handle_unknown="ignore"), ["sex", "embarked"]), remainder="passthrough"),
        HistGradientBoostingClassifier(random_state=0),
    ),
}
scores = {name: cross_val_score(model, X_train_model, y_train, cv=10) for name, model in candidates.items()}
pd.DataFrame({"mean accuracy": {name: s.mean() for name, s in scores.items()},
              "standard deviation across folds": {name: s.std() for name, s in scores.items()}}).round(3)
```

The four ways of filling the age give the logistic regression almost the same accuracy, within a point of each other, while the accuracy of a single pipeline varies by two or three points from one fold to the next: on this table, neither the flag nor a fill from the other columns helps this model. Gradient boosting, which uses the gaps directly, scores a little higher, by less than its own variation across folds.

That is a finding about this table, not a rule: with more gaps, or gaps more closely related to the target, the choice can matter much more. Measure it, as here, inside the pipeline.
